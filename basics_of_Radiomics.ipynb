# **Basics of PyRadiomics**

**🧠 What Is PyRadiomics?**

PyRadiomics is a Python library used to extract quantitative features (called radiomic features) from medical images such as CT, MRI, and PET.

Radiomics converts images into numbers that describe:
*   Texture
*   Shape
*   Intensity patterns

These features are then used for machine learning, prognosis, and diagnosis.

**🧩 Radiomics Workflow**

*   Medical Image (CT / MRI)
*   Segmentation Mask (ROI – tumor or organ)
*   Feature Extraction (PyRadiomics)
*   Machine Learning / Statistics

```python
import SimpleITK as sitk

image = sitk.ReadImage('//content/BraTS-GLI-00000-000_0000.nii.gz')
mask  = sitk.ReadImage('/content/BraTS-GLI-00000-000.nii.gz')

print("Image size:", image.GetSize())
print("Mask size:", mask.GetSize())
```

```python
from radiomics import featureextractor

# Create feature extractor
extractor = featureextractor.RadiomicsFeatureExtractor()

# Enable all features
extractor.enableAllFeatures()

mask = mask > 0

# Extract features
features = extractor.execute(image, mask)

# Print first 10 features
for i, (key, value) in enumerate(features.items()):
    print(key, value)
```
